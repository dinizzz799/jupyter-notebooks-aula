### Questão 1: Orçamento de viagem (1,0 ponto)

Escreva um programa que calcule o custo de uma viagem de carro.

O programa deve ler do usuário, usando `input`:

- a distância da viagem, em km (`float`);
- o consumo médio do carro, em km/litro (`float`);
- o preço do litro do combustível, em reais (`float`);
- o valor total dos pedágios, em reais (`float`);
- o número de pessoas que dividirão a despesa (`int`).

Em seguida o programa deve calcular e exibir:

1. a quantidade de litros necessária (distância dividida pelo consumo);
2. o custo do combustível;
3. o custo total (combustível + pedágios);
4. o custo por pessoa;
5. o tipo (`type`) das variáveis da distância e do número de pessoas.

Todos os valores em reais devem ser exibidos com **2 casas decimais** (use `round` ou formatação de string).

Exemplo:

```
Distância da viagem (km): 450
Consumo médio (km/l): 12.5

Total de pedágios (R$): 32.40

Litros necessários: 36.0
Custo do combustível: R$ 212.04
Custo total da viagem: R$ 244.44
Custo por pessoa: R$ 81.48
Tipo da distância: <class 'float'> | Tipo do número de pessoas: <class 'int'>
```

````python
# Entrada de dados com conversão de tipos
distancia_km = float(input("Distância da viagem (km): "))
consumo_kml = float(input("Consumo médio (km/l): "))
preco_litro = float(input("Preço do litro (R$): "))
total_pedagios = float(input("Total de pedágios (R$): "))
numero_pessoas = int(input("Número de pessoas: "))

# Realização dos calculos
litros_necessarios = distancia_km / consumo_kml
custo_combustivel = litros_necessarios * preco_litro
custo_total = custo_combustivel + total_pedagios
custo_por_pessoa = custo_total / numero_pessoas

# Exibição dos resultados formatados
print()
print("Litros necessários: {litros_necessarios:.1f}")
print("Custo do combustível: R$ {custo_combustivel:.2f}")
print("Custo total da viagem: R$ {custo_total:.2f}")
print("Custo por pessoa: R$ {custo_por_pessoa:.2f}")
print("Tipo da distância: {type(distancia_km)} | Tipo do número de pessoas: {type(numero_pessoas)}")
````

1. leia as coordenadas de dois pontos do plano: `x1`, `y1`, `x2` e `y2` (todos `float`);
2. calcule `dx` e `dy` como as diferenças **em valor absoluto** entre as coordenadas;
3. calcule a distância entre os pontos com a fórmula abaixo:

$$distancia = \sqrt{dx^2 + dy^2}$$

4. calcule a área do círculo que tem essa distância como raio;
5. exiba `dx`, `dy`, a distância e a área, todos arredondados para 2 casas decimais com `round`.

Exemplo:

```
x1: 1
y1: 2
x2: 4
y2: 6

dx = 3.0
dy = 4.0
Distância entre os pontos: 5.0
Área do círculo de raio 5.0: 78.54
```

````python
import math

# Entrada das coordenadas cartesianas
x1 = float(input("x1: "))
y1 = float(input("y1: "))
x2 = float(input("x2: "))
y2 = float(input("y2: "))

# Cálculos de diferenças absolutas, distância e área
dx = abs(x2 - x1)
dy = abs(y2 - y1)
distancia = math.sqrt(dx**2 + dy**2)
area_circulo = math.pi * (distancia**2)

# Exibição dos resultados arredondados com round()
print()
print("dx = {round(dx, 2)}")
print("dy = {round(dy, 2)}")
print("Distância entre os pontos: {round(distancia, 2)}")
print("Área do círculo de raio {round(distancia, 2)}: {round(area_circulo, 2)}")
````

### Questão 4: Encapsulamento e refatoração (1,0 ponto)

O código abaixo calcula e exibe a média de três alunos. Ele **funciona**, mas repete a mesma lógica três vezes:

```python
nota1 = 7.0
nota2 = 5.5
nota3 = 9.0
media = (nota1 + nota2 + nota3) / 3
print("Média de Ana:", round(media, 2))

nota1 = 4.0
nota2 = 6.5
nota3 = 5.0
media = (nota1 + nota2 + nota3) / 3
print("Média de Bruno:", round(media, 2))

nota1 = 9.5
nota2 = 8.0
nota3 = 10.0
media = (nota1 + nota2 + nota3) / 3
print("Média de Carla:", round(media, 2))
```

Sua tarefa:

**(a)** **Encapsule** essa lógica em uma função chamada `exibir_media(nome, nota1, nota2, nota3)`, que calcula a
média das três notas e exibe a mensagem no mesmo formato. A função deve ter uma **docstring** descrevendo o que
ela faz e o que significa cada parâmetro.

**(b)** Chame a função três vezes para reproduzir exatamente a mesma saída do código original.

**(c)** Na célula de texto indicada, explique com suas palavras: o que é **encapsulamento**, o que é
**refatoração**, e por que a versão com função é melhor que a original.

Saída esperada:

```
Média de Ana: 7.17
Média de Bruno: 5.17
Média de Carla: 9.17
```


````python
# (a) Definição da função de cálculo de média
def exibir_media(nome_aluno, nota1, nota2, nota3):
    """
    Calcula a média aritmética de três notas de um aluno e exibe o resultado formatado.

    Parâmetros:
    nome_aluno (str): O nome do estudante.
    nota1 (float): A primeira nota obtida.
    nota2 (float): A segunda nota obtida.
    nota3 (float): A terceira nota obtida.
    """
    media_final = (nota1 + nota2 + nota3) / 3
    print("Média de {nome_aluno}: {round(media_final, 2)}")


# (b) Chamadas da função para reproduzir a saída desejada
exibir_media("Ana", 7.0, 5.5, 9.0)
exibir_media("Bruno", 4.0, 6.5, 5.0)
exibir_media("Carla", 9.5, 8.0, 10.0)
````

Encapsulamento é a ideia de colocar um conjunto de instruções e variáveis dentro de uma estrutura isolada, como uma função. Quem usa essa função só precisa saber quais dados entrarão (os parâmetros) e o que sairá (o resultado), sem precisar entender como tudo funciona por dentro. Isso torna o código mais organizado e mais fácil de usar.

Refatoração é o ato de reorganizar o código sem mudar o que ele faz. O objetivo é deixar o código mais limpo, mais eficiente e mais fácil de entender, mesmo que o resultado final continue o mesmo.

A versão refatorada é melhor por três motivos principais:

Primeiro, elimina a duplicação de código. O princípio DRY  "Don’t Repeat Yourself"  diz que não se deve repetir a mesma coisa várias vezes. A versão atual evita isso, já que a fórmula de cálculo e o print não aparecem repetidos por todo o código.

Segundo, facilita a manutenção. Se, por exemplo, no futuro precisar mudar o cálculo da média  como adicionar uma 4ª nota ou usar pesos diferentes , basta alterar o código dentro da função. Não é preciso procurar e corrigir cada trecho onde a média é calculada. Isso reduz erros e economiza tempo.

Terceiro, melhora a clareza. O programa principal fica mais direto, mais leve e mais fácil de ler. A função desempenha uma tarefa específica, e o restante do código se concentra no que precisa ser feito, sem detalhes complexos. Isso faz com que o código como um todo seja mais compreensível e mais profissional.

**(a)** Escreva uma função chamada `moldura(texto, caractere, largura)` que exibe um texto dentro de uma moldura.
A função deve:

- imprimir uma linha formada pela repetição de `caractere`, com a `largura` especificada;
- imprimir o `texto` entre dois `caractere` (um no início e outro no fim da linha);
- imprimir novamente a linha de fechamento.

A função deve ter uma **docstring**. Note que ela **não retorna valor**, apenas exibe (retorna `None`).

Exemplo:

```
moldura("Trabalho 1", "*", 20)
```

Saída:

```
********************
* Trabalho 1
********************
```

**(b)** Chame `moldura` **três vezes**, com textos, caracteres e larguras diferentes, mostrando que a função foi
**generalizada** pelos parâmetros.

````python
# (a) Função para desenhar a moldura ao redor do texto
def moldura(texto, caractere, largura):
    """
    Exibe uma mensagem envolvida por uma moldura retangular personalizável.

    Parâmetros:
    texto (str): O texto a ser exibido dentro da moldura.
    caractere (str): O caractere utilizado para desenhar a borda.
    largura (int): O tamanho total da largura da moldura em caracteres.
    """
    linha_borda = caractere * largura

    # Cálculo da quantidade de espaços para alinhar o caractere final na largura correta
    quantidade_espacos = largura - len(texto) - 3
    linha_conteudo = caractere + " " + texto + (" " * quantidade_espacos) + caractere

    print(linha_borda)
    print(linha_conteudo)
    print(linha_borda)


# (b) Chamadas demonstrando a generalização com parâmetros variados
moldura("Trabalho 1", "*", 20)
print()
moldura("Aviso Importante", "!", 35)
print()
moldura("Python", "=", 15)
````

```
2000 é bissexto? True
...
eh_triangulo(3, 4, 5) -> True
eh_triangulo(1, 2, 10) -> False
eh_triangulo(-1, 4, 5) -> False
```

````python
# (a) Função para verificar se um ano é bissexto
def eh_bissexto(ano):
    """
    Verifica se um ano informado é bissexto.

    Parâmetros:
    ano (int): O ano a ser verificado.

    Retorno:
    bool: True se o ano for bissexto, False caso contrário.
    """
    return (ano % 4 == 0) and (not (ano % 100 == 0) or (ano % 400 == 0))


# (b) Função para verificar a existência de um triângulo
def eh_triangulo(lado_a, lado_b, lado_c):
    """
    Verifica se três medidas podem formar um triângulo válido.

    Parâmetros:
    lado_a (float/int): Medida do primeiro lado.
    lado_b (float/int): Medida do segundo lado.
    lado_c (float/int): Medida do terceiro lado.

    Retorno:
    bool: True se os lados formarem um triângulo, False caso contrário.
    """
    sao_positivos = lado_a > 0 and lado_b > 0 and lado_c > 0
    desigualdade_triangular = (lado_a < lado_b + lado_c) and (lado_b < lado_a + lado_c) and (lado_c < lado_a + lado_b)
    return sao_positivos and desigualdade_triangular


# (c) Verificação do período de 1996 a 2005
for ano in range(1996, 2006):
    print("{ano} é bissexto? {eh_bissexto(ano)}")

print()

# (d) Testes da função eh_triangulo
print("eh_triangulo(3, 4, 5) -> {eh_triangulo(3, 4, 5)}")
print("eh_triangulo(1, 2, 10) -> {eh_triangulo(1, 2, 10)}")
print("eh_triangulo(-1, 4, 5) -> {eh_triangulo(-1, 4, 5)}")
````

### Questão 8: Condicionais encadeadas e aninhadas (1,0 ponto)

Escreva uma função chamada `classificar_imc(peso, altura)` que:

1. **valide os dados**: se `peso` **ou** `altura` for menor ou igual a zero, a função deve
   retornar `None`, sem calcular nada;
2. caso os dados sejam válidos, calcule o IMC com a fórmula `imc = peso / altura ** 2`;
3. use uma **condicional encadeada** (`if`/`elif`/`else`) para **retornar** a classificação correspondente:

| IMC | Classificação |
|---|---|
| menor que 18,5 | `"Abaixo do peso"` |
| de 18,5 a 24,9 | `"Peso normal"` |
| de 25,0 a 29,9 | `"Sobrepeso"` |
| 30,0 ou mais | `"Obesidade"` |

Depois, escreva um programa que leia o peso e a altura do usuário, chame a função e:

- se o retorno for `None`, exiba a mensagem `"Dados inválidos: peso e altura devem ser positivos."`;
- caso contrário, exiba o IMC arredondado com 2 casas decimais e a classificação.

A função deve ter **docstring**.

Exemplo:

```
Peso (kg): 72.5
Altura (m): 1.75
IMC: 23.67 -> Peso normal
```

```
Peso (kg): 70
Altura (m): 0
Dados inválidos: peso e altura devem ser positivos.
```

````python
def classificar_imc(peso, altura):
    """
    Valida o peso e a altura, calcula o IMC e retorna a sua classificação.

    Parâmetros:
    peso (float): O peso em quilogramas.
    altura (float): A altura em metros.

    Retorno:
    str ou None: Retorna a string com a classificação do IMC ou None se os dados forem inválidos.
    """
    # 1. Validação dos dados de entrada
    if peso <= 0 or altura <= 0:
        return None

    # 2. Cálculo do IMC
    imc = peso / (altura ** 2)

    # 3. Classificação com condicional encadeada
    if imc < 18.5:
        return "Abaixo do peso"
    elif imc <= 24.9:
        return "Peso normal"
    elif imc <= 29.9:
        return "Sobrepeso"
    else:
        return "Obesidade"


# Programa principal: leitura dos dados do usuário
peso_informado = float(input("Peso (kg): "))
altura_informada = float(input("Altura (m): "))

# Chamada da função
classificacao = classificar_imc(peso_informado, altura_informada)

# Exibição da mensagem com base no retorno
if classificacao is None:
    print("Dados inválidos: peso e altura devem ser positivos.")
else:
    imc_calculado = peso_informado / (altura_informada ** 2)
    print("IMC: {round(imc_calculado, 2)} -> {classificacao}")
````

````python
def media_notas(nota1, nota2, nota3):
    """Calcula e retorna a média aritmética de três notas."""
    return (nota1 + nota2 + nota3) / 3


def situacao(media):
    """
    Retorna a situação do aluno com base na média.

    Parâmetros:
    media (float): A média obtida pelo aluno.

    Retorno:
    str: 'Aprovado' (>=7.0), 'Recuperação' (>=4.0 e <7.0) ou 'Reprovado' (<4.0).
    """
    if media >= 7.0:
        return "Aprovado"
    elif media >= 4.0:
        return "Recuperação"
    else:
        return "Reprovado"


# Testes com valores fixos
m1 = media_notas(7.0, 5.5, 9.0)
m2 = media_notas(4.0, 3.0, 2.0)
m3 = media_notas(9.5, 8.0, 10.0)

print(f"Ana: {m1:.1f} - {situacao(m1)}")
print(f"Bruno: {m2:.1f} - {situacao(m2)}")
print(f"Carla: {m3:.1f} - {situacao(m3)}")
````

````python
def media_notas(nota1, nota2, nota3):
    """Calcula e retorna a média aritmética de três notas."""
    return (nota1 + nota2 + nota3) / 3


def situacao(media):
````
````python
    """
    Retorna a situação do aluno com base na média.

    Parâmetros:
    media (float): A média calculada do aluno.

    Retorno:
    str: 'Aprovado', 'Recuperação' ou 'Reprovado'.
    """
    if media >= 7.0:
        return "Aprovado"
    elif media >= 4.0:
        return "Recuperação"
    else:
        return "Reprovado"


total_alunos = int(input("Quantos alunos? "))

boletim = []
soma_medias = 0.0
aprovados = 0
maior_media = -1.0
melhor_aluno = ""

for i in range(1, total_alunos + 1):
    print()
    nome = input(f"Aluno {i} - nome: ")
    n1 = float(input("Nota 1: "))
    n2 = float(input("Nota 2: "))
    n3 = float(input("Nota 3: "))

    media_aluno = media_notas(n1, n2, n3)
    situacao_aluno = situacao(media_aluno)

    boletim.append((nome, media_aluno, situacao_aluno))

    # Atualização de estatísticas
    soma_medias += media_aluno
    if situacao_aluno == "Aprovado":
        aprovados += 1
    if media_aluno > maior_media:
        maior_media = media_aluno
        melhor_aluno = nome

# Exibição do Boletim
print("BOLETIM")
for nome, med, sit in boletim:
    print(f"{nome}: {med:.1f} - {sit}")


# Cálculos Finais
media_geral = soma_medias / total_alunos
percentual_aprovados = (aprovados / total_alunos) * 100

# Exibição das Estatísticas
print(f"Média geral da turma: {media_geral:.2f}")
print(f"Aprovados: {aprovados} de {total_alunos} ({percentual_aprovados:.2f}%)")
print(f"Melhor média: {melhor_aluno} ({maior_media:.2f})")
````
