#### Import do módulo `jupyturtle`

````python
from os.path import basename, exists

def download(url):
    filename = basename(url)
    if not exists(filename):
        from urllib.request import urlretrieve

        local, _ = urlretrieve(url, filename)
        print("Download realizado: " + str(local))
    return filename

download('https://github.com/ramalho/jupyturtle/releases/download/2024-03/jupyturtle.py');
````

### Exercício 1

Escreva uma função chamada `rectangle` que desenhe um retângulo com os comprimentos de lado especificados. Por exemplo, aqui está um retângulo com `80` unidades de largura e `40` unidades de altura.

Exemplo:

```
make_turtle()
rectangle(80, 40)
```

Dica: não esqueça de importar o módulo `jupyturtle` antes de executar seu código

````python
from jupyturtle import make_turtle, forward, right

def rectangle(width, height):
    Desenha um retângulo com a largura (width) e altura (height) especificadas.
    for _ in range(2):
        forward(width)
        right(90)
        forward(height)
        right(90)
````

Escreva uma função chamada `losango` que desenhe um losango com um determinado comprimento de lado e um determinado ângulo interno. Por exemplo, aqui está um losango com comprimento de lado 50 e ângulo interno de 60 graus.

### Exercício 3

Escreva um programa que simule uma partida de Pedra, Papel e Tesoura entre o usuário e o computador.

O programa deve ler a jogada do usuário (rock, paper ou scissors), sortear a jogada do computador entre as mesmas opções e exibir o resultado da partida (vitória, derrota ou empate), seguindo as regras clássicas: pedra vence tesoura, tesoura vence papel, papel vence pedra.

Exemplo:

```
Escolha pedra, papel ou tesoure: pedra
Você escolheu: pedra
Computador escolheu: tesoura
Você venceu! 🎉
```

Dica: use `random.choice` para sortear a jogada do computador e combine as comparações de vitória em uma única estrutura `if/elif/else` com os operadores `and/or`.

````python
import random

def jokenpo():
    opcoes = ["pedra", "papel", "tesoura"]
    usuario = input("Escolha pedra, papel ou tesoura: ").strip().lower()
    if usuario not in opcoes:
        print("Opção inválida! Escolha entre pedra, papel ou tesoura.")
        return
    computador = random.choice(opcoes)
    print("Você escolheu: {usuario}")
    print("Computador escolheu: {computador}")
    if usuario == computador:
        print("Empate! 🤝")
    elif (usuario == "pedra" and computador == "tesoura") or \
         (usuario == "tesoura" and computador == "papel") or \
         (usuario == "papel" and computador == "pedra"):
        print("Você venceu! 🎉")
    else:
        print("Você perdeu! 😢")
````

### Exercício 4

Escreva um programa que aplique um pequeno quiz de 3 perguntas ao usuário, contabilizando os acertos em uma variável `score`.

Perguntas e respostas esperadas:

```
Qual a capital da França? → Paris
O que a sigla CPU significa? → deve conter a expressão Central Processing Unit
Qual a linguagem de programação utilizada na disciplina? → Python
```

Para cada pergunta, leia a resposta do usuário, compare (ignorando espaços extras e diferenças entre maiúsculas/minúsculas) e exiba "Correto!" ou a mensagem informando a resposta correta. Ao final, exiba a pontuação total.

Exemplo:

```
Qual a capital da França? paris
Correto!
O que a sigla CPU significa? central processing unit
Correto!
Qual a linguagem de programação utilizada na disciplina? java
Incorreta. A resposta é Python.

Seu score final: 2/3
```

Dica: use `.strip()` e `.lower()` para normalizar a entrada, `==` para comparações exatas e o operador `in` para verificar se uma expressão está contida na resposta.

````python
def aplicar_quiz():
    score = 0
    total_perguntas = 3
    resp1 = input("Qual a capital da França? ").strip().lower()
    if resp1 == "paris":
        print("Correto!")
        score += 1
    else:
        print("Incorreta. A resposta é Paris.")
    resp2 = input("O que a sigla CPU significa? ").strip().lower()
    if "central processing unit" in resp2:
        print("Correto!")
        score += 1
    else:
        print("Incorreta. A resposta deve conter 'Central Processing Unit'.")
    resp3 = input("Qual a linguagem de programação utilizada na disciplina? ").strip().lower()
    if resp3 == "python":
        print("Correto!")
        score += 1
    else:
        print("Incorreta. A resposta é Python.")
    print("\nSeu score final: {score}/{total_perguntas}")
````

### Exercício 5

Escreva um programa que sorteie uma palavra embaralhada e peça ao usuário para adivinhar a palavra original, repetindo a pergunta até que ele acerte ou digite `quit` para desistir.

Use a palavra `python`, embaralhe suas letras e exiba o resultado para o usuário. Em seguida, entre em um laço que lê a tentativa do usuário: se for `quit`, exiba a palavra correta e encerre; se for igual à palavra original, exiba uma mensagem de acerto e encerre; caso contrário, informe que a tentativa está errada e peça uma nova.

Exemplo:
```
Desembaralhe esta palavra: thopyn
Sua tentativa (ou digite 'quit' para sair): tyhon
Não foi dessa vez, tente novamente!
Sua tentativa (ou digite 'quit' para sair): python
🎉 Você acertou!
```

Dica: use `random.sample` para embaralhar as letras da palavra e `''.join()` para transformar o resultado de volta em string; use `while True` com `break` para controlar a repetição.

````python
import random
def jogo_desembaralhar():
    palavra_original = "python"
    letras = random.sample(palavra_original, len(palavra_original))
    palavra_embaralhada = "".join(letras)
    print("Desembaralhe esta palavra: {palavra_embaralhada}")
    while True:
        tentativa = input("Sua tentativa (ou digite 'quit' para sair): ").strip().lower()
        if tentativa == "quit":
            print("Você desistiu! A palavra correta era: {palavra_original}")
            break
        elif tentativa == palavra_original:
            print("🎉 Você acertou!")
            break
        else:
            print("Não foi dessa vez, tente novamente!")
````

### Exercício 6

Desenvolva um programa que selecione um número aleatório entre 0 e 100. O usuário deverá tentar adivinhar esse número, recebendo uma dica ("maior" ou "menor") a cada tentativa incorreta. Ao final, o programa exibirá a sequência de palpites do usuário e o total de tentativas feitas até acertar o número correto.

Exemplo:
```
Digite um número: 50
O número correto é menor que este
Digite um número: 25
O número correto é maior que este
Digite um número: 37
Parabéns você acertou!
Os seus chutes foram: 50 25 37
O total de chutes foram:  3
```

Dica: use `random.randint` para sortear o número, uma lista para armazenar os palpites e um laço `while True` com `break/continue` para controlar as tentativas.

````python
import random
def jogo_adivinhacao():
    numero_secreto = random.randint(0, 100)
    palpites = []
    while True:
        try:
            chute = int(input("Digite um número: "))
        except ValueError:
            print("Por favor, digite apenas números inteiros válidos.")
            continue
        palpites.append(chute)
        if chute == numero_secreto:
            print("Parabéns você acertou!")
            break
        elif chute < numero_secreto:
            print("O número correto é maior que este")
        else:
            print("O número correto é menor que este")
    chutes_str = " ".join(str(p) for p in palpites) + " "
    print("Os seus chutes foram: {chutes_str}")
    print("O total de chutes foram:  {len(palpites)}")
````
